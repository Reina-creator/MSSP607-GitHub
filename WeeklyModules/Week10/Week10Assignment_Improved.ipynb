This notebook uses a lambda function to return a new DataFrame containing students whose total scores are **at or above the average**. The original DataFrame is preserved.

## 2. Load and check the data

Check the file path, required column, and score values before filtering. These checks stop with a clear message if the input is unsuitable; they do not change student scores or remove rows.

```python
data_file = repo_path / "data" / "Student Performance Data.csv"

if not data_file.is_file():
    raise FileNotFoundError(f"Data file not found: {data_file}")

students = pd.read_csv(data_file)

if "total score" not in students.columns:
    raise ValueError("The dataset must contain a total score column.")

if students.empty:
    raise ValueError("The dataset contains no student records.")

if not pd.api.types.is_numeric_dtype(students["total score"]):
    raise ValueError("The total score column must contain numeric values.")

if students["total score"].isna().any():
    raise ValueError("The total score column contains missing values. Review them before filtering.")

if students["total score"].isin([float("inf"), float("-inf")]).any():
    raise ValueError("The total score column contains infinite values. Review them before filtering.")

print("Dataset shape (rows, columns):", students.shape)
print("Missing total scores:", students["total score"].isna().sum())
print("Total score data type:", students["total score"].dtype)

students.head()
```

## 3. Define the lambda function

`>=` includes scores equal to the average. `.loc` selects the qualifying rows, and `.copy()` returns a separate DataFrame. The function name reflects that the selection includes the average itself.

```python
at_or_above_average = lambda data: data.loc[
    data["total score"] >= data["total score"].mean()
].copy()
```

## 4. Apply the function and review the result

The preview shows the first five selected students. The summary confirms the average, record counts, and returned object type.

```python
above_average_students = at_or_above_average(students)

above_average_students.head()
```

```python
print("Average total score:", students["total score"].mean())
print("Original number of students:", len(students))
print("Number of selected students:", len(above_average_students))
print("Returned object type:", type(above_average_students))
```
