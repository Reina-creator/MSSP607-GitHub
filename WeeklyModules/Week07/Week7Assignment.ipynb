```python
import pandas as pd
import matplotlib.pyplot as plt

sales = pd.read_csv(
    "/content/MSSP607-GitHub/data/InternetSales.csv",
    encoding="latin1",
    low_memory=False
)

sales.head()
```

```python
sales["OrderDate"] = pd.to_datetime(sales["OrderDate"])
sales["Year"] = sales["OrderDate"].dt.year
yearly_sales = sales.groupby("Year")["SalesAmount"].sum()
```

```python
plt.figure(figsize=(8, 5))
yearly_sales.plot(kind="bar", color="steelblue")
plt.title("Yearly Internet Sales")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("yearly_sales.pdf", format="pdf", bbox_inches="tight")
plt.show()
```
