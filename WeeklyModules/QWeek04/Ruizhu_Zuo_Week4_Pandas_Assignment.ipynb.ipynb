```python
import pandas as pd

earthquakes = pd.read_csv("parsed.csv")

earthquakes.head()
```

```python
earthquakes.columns
```

```python
#Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
japan_mb = earthquakes[
    (earthquakes["parsed_place"] == "Japan") &
    (earthquakes["magType"] == "mb")
]

japan_95th_percentile = japan_mb["mag"].quantile(0.95)

print("95th percentile:", japan_95th_percentile)
```

```python
# Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
indonesia = earthquakes[
    earthquakes["parsed_place"] == "Indonesia"
]

indonesia_tsunami_percentage = indonesia["tsunami"].mean() * 100

print(
    f"Percentage of earthquakes in Indonesia with tsunamis: "
    f"{indonesia_tsunami_percentage:.2f}%"
)
```

```python
#Get summary statistics for earthquakes in Nevada
nevada = earthquakes[
    earthquakes["parsed_place"] == "Nevada"
]

nevada.describe()
```

```python
#Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire. Use Bolivia, Chile, Ecuador, Peru, Costa Rica, Guatemala, Mexico (be sure not to select New Mexico), Japan, Philippines, Indonesia, New Zealand, Antarctica (look for the Antarctic), Canada, Fiji, Alaska, Washington, California, Russia, Taiwan, Tonga, and the Kermadec Islands).
ring_of_fire_locations = [
    "Bolivia", "Chile", "Ecuador", "Peru",
    "Costa Rica", "Guatemala", "^Mexico",
    "Japan", "Philippines", "Indonesia",
    "New Zealand", "Antarctic", "Canada",
    "Fiji", "Alaska", "Washington",
    "California", "Russia", "Taiwan",
    "Tonga", "Kermadec Islands"
]

pattern = "|".join(ring_of_fire_locations)

earthquakes["ring_of_fire"] = earthquakes[
    "parsed_place"
].str.contains(
    pattern,
    case=False,
    na=False
)

earthquakes[["parsed_place", "ring_of_fire"]].head()
```

```python
#Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them
ring_of_fire_counts = earthquakes[
    "ring_of_fire"
].value_counts()

print(ring_of_fire_counts)
```

```python
#Find the tsunami count along the Ring of Fire.
ring_of_fire_tsunami_count = earthquakes.loc[
    earthquakes["ring_of_fire"],
    "tsunami"].sum()

print("Tsunami count along the Ring of Fire:",
    ring_of_fire_tsunami_count)
```
