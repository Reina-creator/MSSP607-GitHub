This notebook uses a lambda function to take the Student Performance DataFrame and return a new DataFrame containing students whose total scores are greater than or equal to the average total score.

```python
import pandas as pd

students = pd.read_csv("Student Performance Data.csv")

students.head()
```

```python
print("Number of rows and columns:", students.shape)
print("Average total score:", students["total score"].mean())
```

```python
above_average = lambda data: data[
    data["total score"] >= data["total score"].mean()
].copy()

above_average_students = above_average(students)

above_average_students.head()
```

```python
print("Returned object type:", type(above_average_students))
print("Number of students:", len(above_average_students))

above_average_students
```
